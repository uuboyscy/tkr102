```python
import pandas as pd
```

```python
file_path = "holding.parquet"
df = pd.read_parquet(file_path)
df
```

```python
"""
select distinct portno
from df;

select count(distinct(portno)) from df;
"""
df["portno"].unique()
```

```python
"""
select
    portno,
    count(1)
from df
group by portno;
"""
df.groupby(["portno"], as_index=False).size()
```

```python
df.groupby(
    ["portno"], as_index=False,
).agg(
    portno_count=("portno", "size"),
    share_sum=("nbr_shares", "sum"),
)
```

```python
"""
select
    portno,
    sum(percent_tna) as perc
from df
group by portno, eff_dt
"""
df.groupby(
    ["portno", "eff_dt"],
    as_index=False,
).agg(
    perc_123=("percent_tna", "sum")
)
```

```python
df.groupby(
    ["portno", "eff_dt"],
    as_index=False,
)["percent_tna"].sum().rename(
    columns={"percent_tna": "new_percent_tna"}
)
```

```python
"""
select
    portno,
    sum(percent_tna) as perc
from df
group by portno, eff_dt
order by portno asc, eff_dt desc;
"""
df.groupby(
    ["portno", "eff_dt"],
    as_index=False,
).agg(
    perc_123=("percent_tna", "sum")
).sort_values(["portno", "eff_dt"], ascending=[True, False])
```
