## Auto commit

```python
from urllib.parse import quote_plus

import pandas as pd
from sqlalchemy import create_engine

username = quote_plus("root")
password = quote_plus("1qaz@WSX")
server = "127.0.0.1:3307"
db_name = "TESTDB"

with create_engine(
    f"mysql+pymysql://{username}:{password}@{server}/{db_name}?autocommit=true",
).connect() as conn:
    # Do something to transform dataframe
    df = pd.read_sql("Staff", conn)
df
```

```python
df["ID"] = df["ID"].str.replace("01", "02")
df
```

```python
df.columns
```

```python
df[
    [
        'ID',
        'DeptId',
        'Gender',
        'recordDt',
        'Name',
        'Age',
        'Salary'
    ]
]
```

```python
with create_engine(
    f"mysql+pymysql://{username}:{password}@{server}/{db_name}?autocommit=true",
).connect() as conn:
    # Do something to transform dataframe
    df[
        [
            'ID',
            'DeptId',
            'Gender',
            'recordDt',
            'Name',
            'Age',
            'Salary'
        ]
    ].to_sql(
        "Staff",
        conn,
        if_exists="append",
        index=False,
    )
```
