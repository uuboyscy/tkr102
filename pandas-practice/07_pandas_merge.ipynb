## SQL UNION

```python
import pandas as pd

df = pd.DataFrame(
    data=[
        ["Allen", 175, 22],
        ["Ted", 180, 25]
    ],
    columns=["Name", "Height", "Age"]
)

df_1 = pd.DataFrame(
    data=[
        ["Jake", 177, 28],
        ["Mary", 170, 24]
    ],
    columns=["Name", "Height", "Age"]
)
```

```python
pd.concat(
    [df, df_1],
    ignore_index=True,
)
```

```python
"""
select * from df
union
select * from df_1
"""
df_concat = pd.concat(
    [df, df_1],
    ignore_index=True,
)
df_concat
```

```python
df_concat.drop_duplicates()
```

```python
# Drop duplicates by specific columns
df_concat.drop_duplicates(["Name", "Height"])
```
